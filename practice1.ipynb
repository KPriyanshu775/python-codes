```python
def con(n):
    if n == 0:
        return
    count = 0
    while n > 0:
        n = n // 10
        count += 1
    return count
con(12345)
```

```python
def fact(n):
    if n == 0 or n == 1:
        return 1
    return fact(n - 1) * n

print(fact(5))
```
