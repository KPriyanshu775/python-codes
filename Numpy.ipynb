```python
import numpy as np

a = np.array(43)
b = np.array([3,2,5,1,5,6])
c = np.array([[1,2,3,4], [5,6,7,8]])
d = np.array([[[3,4,5], [1,2,3], [6,7,8]]])

print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
```
